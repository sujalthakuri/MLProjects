```python
import sklearn
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.shape
```

```python
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:,0:5],df.iloc[:,-1],test_size=0.3,random_state=False)
```

```python
x_train
```

# Column Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers = [
    ('tnf1',SimpleImputer(),['fever']),('tnf2', OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),('tnf3',OneHotEncoder(sparse_output=False,drop="first"),['gender','city'])
], remainder='passthrough')
```

```python
transformer.fit_transform(x_train)
```
